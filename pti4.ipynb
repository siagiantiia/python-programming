```python
def input_nilai():
    nilai = []
    jumlah = int(input("Masukkan jumlah nilai mahasiswa: "))
    for i in range(jumlah):
        n = float(input(f"Masukkan nilai ke-{i+1}: "))
        nilai.append(n)
    return nilai

def rata_rata(nilai):
    if len(nilai) == 0:
        return 0
    return sum(nilai) / len(nilai)

def nilai_tertinggi(nilai):
    if len(nilai) == 0:
        return None
    return max(nilai)

def nilai_terendah(nilai):
    if len(nilai) == 0:
        return None
    return min(nilai)

def tampilkan_hasil(nilai):
    if len(nilai) == 0:
        print("Tidak ada nilai untuk dianalisis.")
        return
    rata = rata_rata(nilai)
    tertinggi = nilai_tertinggi(nilai)
    terendah = nilai_terendah(nilai)
    print("\n=== Hasil Analisis Nilai ===")
    print(f"Jumlah nilai: {len(nilai)}")
    print(f"Rata-rata: {rata:.2f}")
    print(f"Nilai tertinggi: {tertinggi}")
    print(f"Nilai terendah: {terendah}")

nilai_mahasiswa = input_nilai()
tampilkan_hasil(nilai_mahasiswa)
```

```python
def input_data():
    nama = input("Masukkan nama karyawan: ")
    jabatan = input("Masukkan jabatan (Manager/Supervisor/Staff): ").capitalize() 
    jam = float(input("Masukkan jam kerja: "))
    return nama, jabatan, jam

def hitung_upah_per_jam(jabatan):
    if jabatan == "Manager":
        return 100000
    elif jabatan == "Supervisor":
        return 75000
    elif jabatan == "Staff":
        return 50000
    else:
        print("Jabatan tidak valid. Menggunakan upah default 0.")
        return 0

def hitung_gaji(jabatan, jam):
    upah = hitung_upah_per_jam(jabatan)
    return upah * jam

def tampilkan_slip(nama, jabatan, jam, total):
    print("\n=== Slip Gaji Karyawan ===")
    print(f"Nama: {nama}")
    print(f"Jabatan: {jabatan}")
    print(f"Jam Kerja: {jam}")
    print(f"Total Gaji: Rp{total:,.0f}")

nama, jabatan, jam = input_data()
total_gaji = hitung_gaji(jabatan, jam)
tampilkan_slip(nama, jabatan, jam, total_gaji)
```
