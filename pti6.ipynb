```python
mata_kuliah_tersedia = {"logmat", "daspro", "pti", "op", "agama"}

data_mahasiswa = []

def tambah_mahasiswa():
    nama = input("Masukkan nama mahasiswa: ")
    nim = input("Masukkan NIM: ")

    print("\nMata kuliah yang tersedia:")
    for mk in mata_kuliah_tersedia:
        print(f"- {mk}")
    
    mk_ambil = input("Masukkan mata kuliah yang diambil (pisahkan dengan koma): ")
    daftar_mk = [mk.strip() for mk in mk_ambil.split(",") if mk.strip() in mata_kuliah_tersedia]
   
    mahasiswa = (nama, nim, daftar_mk)
  
    data_mahasiswa.append(mahasiswa)

indeks_mahasiswa = {}

while True:
    print("\n--- Sistem Data Mahasiswa ---")
    print("a. Tambah Mahasiswa")
    print("b. Tampilkan Data Mahasiswa")
    print("c. Keluar")
    
    pilihan = input("Pilih menu (a/b/c): ")
    
    if pilihan == "a":
        tambah_mahasiswa()
        indeks_mahasiswa[data_mahasiswa[-1][1]] = len(data_mahasiswa) - 1
    
    elif pilihan == "b":
        print("\n--- Data Mahasiswa ---")
        if not data_mahasiswa:
            print("Belum ada data.")
        else:
            for idx, mhs in enumerate(data_mahasiswa):
                nama, nim, mk = mhs
                print(f"{idx+1}. Nama: {nama}, NIM: {nim}, Mata Kuliah: {', '.join(mk)}")
    
    elif pilihan == "c":
        print("Terima kasih!")
        break
    
    else:
        print("Pilihan tidak valid.")
```
