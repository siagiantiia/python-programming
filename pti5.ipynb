```python
nama =   input ("masukkan nama:") #tipe string
umur = int ( input  ("masukkan umur:")) #int 
tinggi = float (input ( "masukkan tinggi badan:")) #float
hobi = input ( " masukkan hobi (pisahkan dalam koma ):") . split (",") #list

print ("\n === Data Pengguna ===")
print ( f" nama : { nama} ")
print (f" umur : {umur} tahun ")
print (f" tinggi : {tinggi} cm")
print (" hobi :")
for i , h in enumerate ( hobi , start = 1) :
    print (f"  {i} . {h.strip()}")
    
```
