```python
import pandas as pd


df = pd.read_csv('data penjualan.csv', sep=';')

print("Data Penjualan")
print(df.head())

print("\nJumlah Nilai Kosong:")
print(df.isnull().sum())

df['Harga Satuan'] = (
    df['Harga Satuan']
    .str.replace('Rp', '', regex=False)
    .str.replace('.', '', regex=False)
    .str.strip()
    .astype(int)
)

df['Total Penjualan'] = df['Jumlah Terjual'] * df['Harga Satuan']

print("\nData Setelah Ditambah Kolom Total Penjualan")
print(df.head())
```

```python
import random
bola = ['Merah']*6 + ['Biru']*4 + ['Hijau']*5

percobaan = 1000
warna_sama = 0

for i in range(percobaan):
    ambil = random.sample(bola, 2)

    if ambil[0] == ambil[1]:
        warna_sama += 1

peluang_simulasi = warna_sama / percobaan

print("Jumlah percobaan :", percobaan)
print("Warna sama :", warna_sama)
print("Peluang simulasi :", peluang_simulasi)
```

```python
from scipy.stats import binom

n = 20
p = 0.08

tepat_2 = binom.pmf(2, n, p)

maks_3 = binom.cdf(3, n, p)

print("Peluang tepat 2 produk cacat =", tepat_2)
print("Peluang paling banyak 3 produk cacat =", maks_3)
```

```python
from scipy.stats import norm

mean = 12
std = 3

kurang_10 = norm.cdf(10, mean, std)

antara_10_15 = norm.cdf(15, mean, std) - norm.cdf(10, mean, std)

print(f"Peluang dilayani kurang dari 10 menit = {kurang_10:.4f}")
print(f"Peluang dilayani antara 10 sampai 15 menit = {antara_10_15:.4f}")
```

```python
import numpy as np
from statistics import multimode

data = [70, 75, 80, 68, 90, 85, 72, 88, 76, 95, 60, 78, 82, 87, 73]

print("Mean =", np.mean(data))
print("Median =", np.median(data))

modus = multimode(data)
if len(modus) == len(data):
    print("Modus = Tidak ada")
else:
    print("Modus =", modus)

print("Varians =", np.var(data))
print("Standar Deviasi =", np.std(data))
print("Nilai Minimum =", np.min(data))
print("Nilai Maksimum =", np.max(data))
print("Range =", np.max(data) - np.min(data))
```

```python
import pandas as pd

df = pd.read_csv('customer_shopping_data.csv')

df['total_belanja'] = df['quantity'] * df['price']

total_belanja = df.groupby(
    ['shopping_mall', 'category']
)['total_belanja'].sum()

print("TOTAL BELANJA PER KATEGORI DAN TEMPAT BELANJA")
print(total_belanja)

df['kelompok_umur'] = pd.cut(
    df['age'],
    bins=[0,20,30,40,50,60,100],
    labels=['<20','21-30','31-40','41-50','51-60','>60']
)

rata_pembelian = df.groupby(
    'kelompok_umur'
)['quantity'].mean()

print("\nRATA-RATA JUMLAH PEMBELIAN PER KELOMPOK UMUR")
print(rata_pembelian)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Sample - Superstore.csv', encoding='latin1')

df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Bulan'] = df['Order Date'].dt.to_period('M')

penjualan_bulanan = df.groupby('Bulan')['Sales'].sum()

plt.figure(figsize=(12,5))
plt.plot(penjualan_bulanan.index.astype(str),
         penjualan_bulanan.values,
         marker='o', color='yellow')
plt.title('Tren Penjualan Bulanan')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
from scipy.stats import t
import math

x_bar = 78
mu = 75
s = 8
n = 30

t_hitung = (x_bar - mu) / (s / math.sqrt(n))

t_tabel = t.ppf(0.975, df=n-1)

print("t hitung =", t_hitung)
print("t tabel =", t_tabel)

if abs(t_hitung) > t_tabel:
    print("Tolak H0")
else:
    print("Gagal Tolak H0")
```

```python
from scipy.stats import ttest_ind_from_stats

mean1 = 82
std1 = 7
n1 = 20

mean2 = 77
std2 = 8
n2 = 22

t_stat, p_value = ttest_ind_from_stats(
    mean1=mean1,
    std1=std1,
    nobs1=n1,
    mean2=mean2,
    std2=std2,
    nobs2=n2
)

print("t statistik =", t_stat)
print("p-value =", p_value)

if p_value < 0.05:
    print("Tolak H0")
else:
    print("Gagal Tolak H0")
```

```python
from statsmodels.stats.proportion import proportions_ztest

cacat = 7
total = 100

z_stat, p_value = proportions_ztest(
    count=cacat,
    nobs=total,
    value=0.10,
    alternative='smaller'
)

print("Z statistik =", z_stat)
print("P-value =", p_value)

if p_value < 0.05:
    print("Tolak H0")
else:
    print("Gagal Tolak H0")
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv('student_scores.csv')

X = df[['Hours']]
y = df['Scores']

model = LinearRegression()
model.fit(X, y)

print("Koefisien (b1) =", model.coef_[0])
print("Intercept (b0) =", model.intercept_)
print("R² =", model.score(X, y))
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv('Housing.csv')

X = df[['area', 'bedrooms']]

y = df['price']

model = LinearRegression()
model.fit(X, y)

print("Koefisien Area =", model.coef_[0])
print("Koefisien Bedrooms =", model.coef_[1])
print("Intercept =", model.intercept_)
print("R² =", model.score(X, y))
```

```python
from math import comb

n = 20
p = 0.08

prob_2 = comb(n, 2) * (p ** 2) * ((1 - p) ** (n - 2))

prob_max_3 = sum(
    comb(n, k) * (p ** k) * ((1 - p) ** (n - k)
    ) for k in range(4)
)

print("P(X = 2) =", prob_2)
print("P(X <= 3) =", prob_max_3)
```
