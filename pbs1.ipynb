```python
df = pd.read_csv('StudentPerformanceFactors.csv')

print("INFO DATA")
print(df.info())

print("DESKRIPSI DATA")
print(df.describe())
```

```python
print(" MISSING VALUES ")
print(df.isnull().sum())
```

```python
print("STATISTIK DESKRIPTIF ")
print(df.describe())
df = df.fillna(df.mean(numeric_only=True))

df = df.drop_duplicates()

print(" SETELAH CLEANING ")
print(df.isnull().sum())
```

```python
df.hist(figsize=(10,8), color='red')
plt.suptitle("Distribusi Data")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='Reds')
plt.title("Korelasi Antar Variabel")
plt.show()
```

```python
if 'Hours_Studied' in df.columns and 'Exam_Score' in df.columns:
    plt.scatter(df['Hours_Studied'], df['Exam_Score'] ,color='red')
    plt.xlabel("Hours Studied")
    plt.ylabel("Exam Score")
    plt.title("Hubungan Jam Belajar vs Nilai")
    plt.show()
```

```python
if 'Exam_Score' in df.columns:
    mean = df['Exam_Score'].mean()
    std = df['Exam_Score'].std()
    n = len(df)

    margin_error = 1.96 * (std / np.sqrt(n))
    lower = mean - margin_error
    upper = mean + margin_error

    print("ESTIMASI ")
    print("Mean:", mean)
    print("Confidence Interval 95%:", (lower, upper))
```
