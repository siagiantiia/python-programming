```python
print("=== SELAMAT DATANG DI BIOSKOP PYTHON ===")

# Data awal
harga_tiket = 40000
ppn = 0.1
promo_hari = True
snack_menu = ["Popcorn", "Churros", "Hotdog"]
minuman_menu = ["Cola", "Teh Botol", "Air Mineral"]
kursi_tersedia = ["A1", "A2", "A3", "B1", "B2"]

# Input nama
nama = input("Masukkan nama Anda: ")
if not nama.replace(" ", "").isalpha():
    print("Nama hanya boleh mengandung huruf dan spasi!")
    exit()

# Input umur
umur_input = input("Masukkan umur Anda: ")
if not umur_input.isdigit():
    print("Umur harus angka!")
    exit()
umur = int(umur_input)

# Kategori dan diskon
if umur < 5:
    print("Maaf, usia terlalu muda.")
    exit()
elif umur >= 60:
    diskon = 0.2
    kategori = "Lansia"
elif umur >= 13:
    diskon = 0.1
    kategori = "Dewasa"
else:
    diskon = 0
    kategori = "Anak"

print(f"Halo {nama}, kategori: {kategori}, diskon: {diskon*100}%")

# Pilih kursi
print("\nDaftar kursi tersedia:")
for i, kursi in enumerate(kursi_tersedia, start=1):
    print(f"{i}. {kursi}")

pilihan_kursi = input("Pilih kursi: ").upper()
if pilihan_kursi in kursi_tersedia:
    kursi_tersedia.remove(pilihan_kursi)
else:
    print("Kursi tidak tersedia!")
    exit()

# Pesan snack
pesanan_snack = []
while True:
    print("\nMenu Snack:")
    for idx, snack in enumerate(snack_menu, start=1):
        print(f"{idx}. {snack}")
    pilih_snack = input("Pilih snack (0 untuk selesai): ")

    if pilih_snack == "0":
        break
    elif pilih_snack.isdigit() and 1 <= int(pilih_snack) <= len(snack_menu):
        pesanan_snack.append(snack_menu[int(pilih_snack) - 1])
    else:
        print("Pilihan tidak valid!")

# Pesan minuman
pesanan_minum = []
while True:
    print("\nMenu Minuman:")
    for idx, minum in enumerate(minuman_menu, start=1):
        print(f"{idx}. {minum}")
    pilih_minum = input("Pilih minuman (0 untuk selesai): ")

    if pilih_minum == "0":
        break
    elif pilih_minum.isdigit() and 1 <= int(pilih_minum) <= len(minuman_menu):
        pesanan_minum.append(minuman_menu[int(pilih_minum) - 1])
    else:
        print("Pilihan tidak valid!")

# Jumlah tiket
jumlah_tiket = input("Masukkan jumlah tiket: ")
if not jumlah_tiket.isdigit():
    jumlah_tiket = 1
else:
    jumlah_tiket = int(jumlah_tiket)

# Perhitungan harga
subtotal = harga_tiket * jumlah_tiket
subtotal_diskon = subtotal - (subtotal * diskon)
total = subtotal_diskon + (subtotal_diskon * ppn)

# Rincian pesanan
pesanan = {
    "Nama": nama,
    "Umur": umur,
    "Kategori": kategori,
    "Kursi": pilihan_kursi,
    "Jumlah Tiket": jumlah_tiket,
    "Snack": pesanan_snack,
    "Minuman": pesanan_minum,
    "Total Bayar": total
}

print("\n=== RINCIAN PESANAN ===")
for key, value in pesanan.items():
    print(f"{key.upper()}: {value}")

# Kode booking
kode_booking = nama[:3].upper() + pilihan_kursi + str(umur)
print("Kode Booking:", kode_booking)

# Tambahan string method
print("Nama join:", "-".join(nama))
print("Nama split:", nama.split())
```
