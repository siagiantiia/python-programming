```python
print("menu hitung luas persegi panjang & lingkaran")

def luas_persegi_panjang(a, b):
    return a * b

def luas_lingkaran(r):
    return 3.14 * r * r

while True:
    print("1. persegi panjang")
    print("2. lingkaran")
    
    menu = int(input("pilih menu: "))
    
    if menu == 1:
        a = int(input("masukkan panjang persegi panjang: "))
        b = int(input("masukkan lebar persegi panjang: "))
        print("hasil luas persegi panjang:", luas_persegi_panjang(a, b))
        break
    elif menu == 2:
        r = float(input("masukkan jari-jari lingkaran: "))
        print("hasil luas lingkaran:", luas_lingkaran(r))
        break
    else:
        print("masukkan angka 1 atau 2")
```
